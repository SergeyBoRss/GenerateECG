# Генерация датасета ЭКГ

```python
from dataclasses import replace

from ecg.batch import generate_dataset
from ecg.config import EcgConfig
```

## Общие параметры

```python
HR_RANGE = (65, 95)
RR_STD_MS = 20.0
SEED = 1
SAMPLE_RATE = 1000
WORKERS = 10
SAVE_IMAGES = True
OUTPUT_DIR = "outputs/dataset"

BASE_CONFIG = EcgConfig(
    hr_bpm=HR_RANGE,
    rr_std_ms=RR_STD_MS,
    seed=SEED,
    sample_rate=SAMPLE_RATE,
)
```

## Типы и количество записей

```python
DATASET_TYPES = {
    "normal": {
        "count": 100,
        "parameters": {},
    },
    "short_pq": {
        "count": 150,
        "parameters": {
            "no_p": 0.0,
            "pq_ms": (80, 120),
            "p_ms": (50, 70),
        },
    },
    "long_pq": {
        "count": 100,
        "parameters": {
            "no_p": 0.0,
            "pq_ms": (200, 300),
        },
    },
    "short_qt": {
        "count": 100,
        "parameters": {"qtc_ms": (300, 340), "t_ms": (100, 130)},
    },
    "long_qt": {
        "count": 100,
        "parameters": {"qtc_ms": (420, 450), "pq_ms": (135, 140)},
    },
    "hr_30_50": {
        "count": 100,
        "parameters": {"hr_bpm": (30, 50)},
    },
    "hr_100_140": {
        "count": 150,
        "parameters": {
            "hr_bpm": (100, 140),
        },
    },
    "wide_qrs": {
        "count": 100,
        "parameters": {"qrs_ms": (120, 200), "t_ms": (70, 90)},
    },
    "rr_std_100": {
        "count": 100,
        "parameters": {"rr_std_ms": 100.0},
    },
}
```

## Генерация

```python
configs = {
    name: replace(BASE_CONFIG, **spec["parameters"])
    for name, spec in DATASET_TYPES.items()
}

for name, config in configs.items():
    count = DATASET_TYPES[name]["count"]
    print(f"\n{name}: {count} записей; HR={config.hr_bpm}; seed={config.seed}")
    meta_path = generate_dataset(
        config,
        OUTPUT_DIR,
        count=count,
        workers=WORKERS,
        save_images=SAVE_IMAGES,
    )
    print(f"Таблица: {meta_path}")
```
